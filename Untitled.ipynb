All data are **ILLUSTRATIVE TEACHING DATA - NOT ACTUAL PBS OBSERVATIONS**. Prices in PKR per kg (illustrative).

```python
import hashlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

FILE =r"C:\Users\USER\Downloads\teaching_prices_clean.csv"
print("SHA-256:", hashlib.sha256(open(FILE, "rb").read()).hexdigest())

df = pd.read_csv(FILE, parse_dates=["date"])
print(df.shape)
print(df["flag"].value_counts(dropna=False))
df[df["flag"].notna()][["date", "item", "city", "price_raw", "flag", "price_clean"]]
```

## 1. Which rows are used?
`price_clean` is the Week 2 cleaned price. It is empty for 4 flagged rows (missing price, the 1,250 outlier, the 9999 placeholder, and the mislabelled `oil b` row that looks like Staple A). The mislabelled row has no valid item, so it drops out of every item. I do not recode it to Staple A without a source.

```python
clean = df.dropna(subset=["price_clean"]).copy()
clean = clean[clean["item"] != "oil b"]
print(clean.groupby("item").size())
```

## 2. Mean, median, spread, and first-to-last change per item

```python
stats = (clean.groupby("item")["price_clean"]
         .agg(n="count", mean="mean", median="median", sd="std", min="min", max="max"))
stats["mean_minus_median"] = stats["mean"] - stats["median"]

# First-to-last change is computed within each item-city series (first and last
# available observation), then averaged over the two cities. Pooling both cities
# by date would mix in a missing Pulse C City 2 value on 2026-01-01.
rows = []
for (item, city), g in clean.groupby(["item", "city"]):
    g = g.sort_values("date")
    first, last = g.iloc[0], g.iloc[-1]
    rows.append({"item": item, "city": city,
                 "first_date": first["date"].date(), "first": first["price_clean"],
                 "last_date": last["date"].date(), "last": last["price_clean"],
                 "pct_change": (last["price_clean"] / first["price_clean"] - 1) * 100})
chg_city = pd.DataFrame(rows)
chg_item = chg_city.groupby("item")["pct_change"].mean().rename("pct_change_first_to_last")

table = stats.join(chg_item).round(2)
table
```

```python
chg_city.round(2)
```

### Why the pooled shortcut misleads
If both cities are averaged by date first, Pulse C starts from City 1 alone on 2026-01-01 (City 2 is missing), so the start is 200.0 instead of a two-city average.

```python
pooled = clean.groupby(["item", "date"])["price_clean"].mean().reset_index()
naive = pooled.sort_values("date").groupby("item")["price_clean"].agg(["first", "last"])
naive["naive_pct_change"] = (naive["last"] / naive["first"] - 1) * 100
naive.join(chg_item).round(2)
```

## 3. With and without a flagged value: Staple A
`price` still holds the raw 1,250 value (City 2, 2026-01-15). `price_clean` leaves it out.

```python
a = df[df["item"] == "Staple A"]
with_flag = a["price"].dropna()
without_flag = a["price_clean"].dropna()

cmp = pd.DataFrame({
    "n": [len(with_flag), len(without_flag)],
    "mean": [with_flag.mean(), without_flag.mean()],
    "median": [with_flag.median(), without_flag.median()],
}, index=["with 1,250", "without 1,250"]).round(2)
cmp.loc["difference"] = cmp.loc["with 1,250"] - cmp.loc["without 1,250"]
cmp
```

```python
# By city: the outlier sits in City 2, so City 1 is untouched
by_city = a.groupby("city").agg(
    mean_with=("price", "mean"), median_with=("price", "median"),
    mean_without=("price_clean", "mean"), median_without=("price_clean", "median")).round(2)
by_city
```

```python
# How far one wild value moves each statistic
shift_mean = with_flag.mean() - without_flag.mean()
shift_median = with_flag.median() - without_flag.median()
print(f"Mean shifts by {shift_mean:.2f} PKR; median shifts by {shift_median:.2f} PKR")
```

### Second example: the 9999 placeholder (Sweetener D)
If the placeholder were treated as a real price, it would behave the same way.

```python
d = df[df["item"] == "Sweetener D"]
raw = pd.to_numeric(d["price_raw"])
pd.DataFrame({"mean": [raw.mean(), d["price_clean"].mean()],
              "median": [raw.median(), d["price_clean"].median()]},
             index=["with 9999", "9999 set to NaN"]).round(2)
```
